# Minimum working example
This notebook runs a minimum working example with bikenetlib.

## Import bikenetlib

The standard way of importing bikenetlib:

```python
import bikenetlib as bnl
```

## Use `bnl.io`

Showcasing `io.slugify()`:

```python
bnl.io.slugify("L'Hospitalet De Llobregat")
```

## Use `bnl.osm`

Showcasing `osm.PBI_CUSTOM_FILTER`:

```python
import osmnx as ox
g = ox.graph_from_place(
    "Frederiksberg municipality",
    custom_filter=bnl.osm.PBI_CUSTOM_FILTER,
)
ox.plot_graph(g, node_size=0);
```

## Use `bnl.utils`

Showcasing `utils.get_principal_bearing()`:

```python
bnl.utils.get_principal_bearing(ox.convert.to_undirected(g))
```
